```python
from matplotlib import colors
import matplotlib.pyplot as plt
#example for bar chart
#data for x axis
x = ['laptop','Mobile','Keyboard','Mouse','Tablet']
#data for y axis
y = [86,59,45,56,95]
#-------------------------------------------------
#bar chart
plt.bar(x,y,color=['red','blue','green','black','blue'])
#label for x-axis
plt.xlabel('Product')
#label for y axis
plt.ylabel('Sales in units')
#-----------------------------------
#title of chart
plt.title('Product Sales(in Unit)')
#-----------------------------------
#Displaying the chart
plt.show()
```
